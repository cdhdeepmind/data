```python
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score, KFold
from sklearn.ensemble import RandomForestClassifier
from tqdm.notebook import tqdm  # JupyterLab 전용 tqdm
import pandas as pd
import numpy as np
import time

# 1. 데이터 로드
iris = load_iris()
X, y = iris.data, iris.target

# 2. 교차 검증 설정 (5번의 테스트를 거침)
kf = KFold(n_splits=5, shuffle=True, random_state=42)
model = RandomForestClassifier(n_estimators=100)

# 3. tqdm을 활용한 반복 학습 진행률 표시
scores = []

# kf.split(X)가 생성하는 반복문에 tqdm을 감싸줍니다.
for train_index, test_index in tqdm(kf.split(X), total=5, desc="교차 검증 진행률"):
    # 데이터 분할
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]
    
    # 모델 학습
    model.fit(X_train, y_train)
    
    # 평가 및 결과 저장
    score = model.score(X_test, y_test)
    scores.append(score)
    
    # 맥북 2017에서 프로그레스 바가 움직이는 것을 확인하기 위해 잠깐 멈춤 (실제 작업 시 삭제 가능)
    time.sleep(0.5)

# 4. 최종 결과 출력
print(f"\n각 회차별 정확도: {np.round(scores, 3)}")
print(f"평균 정확도: {np.mean(scores):.4f}")
```
